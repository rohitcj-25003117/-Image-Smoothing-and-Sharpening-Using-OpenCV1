```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# =========================================================
# Read Image
# =========================================================

image1 = cv2.imread("saveetha.jpg")

image2 = cv2.cvtColor(
    image1,
    cv2.COLOR_BGR2RGB
)
```

```python
# =========================================================
# 1. Smoothing Filter - Averaging Filter
# =========================================================

kernel = np.ones(
    (11, 11),
    np.float32
) / 169

average = cv2.filter2D(
    image2,
    -1,
    kernel
)

plt.figure(figsize=(9, 9))

plt.subplot(1, 2, 1)
plt.imshow(image2)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(average)
plt.title("Average Filter Image")
plt.axis("off")

plt.show()
```

```python
# =========================================================
# 2. Smoothing Filter - Weighted Averaging Filter
# =========================================================

kernel1 = np.array([
    [1, 2, 1],
    [2, 4, 2],
    [1, 2, 1]
]) / 16

weighted_average = cv2.filter2D(
    image2,
    -1,
    kernel1
)

plt.imshow(weighted_average)
plt.title("Weighted Average Filter Image")
plt.axis("off")
plt.show()
```

```python
# =========================================================
# 3. Smoothing Filter - Gaussian Filter
# =========================================================

gaussian_blur = cv2.GaussianBlur(
    image2,
    (33, 33),
    0,
    0
)

plt.imshow(gaussian_blur)
plt.title("Gaussian Blur")
plt.axis("off")
plt.show()
```

```python
# =========================================================
# 4. Smoothing Filter - Median Filter
# =========================================================

median = cv2.medianBlur(
    image2,
    13
)

plt.imshow(median)
plt.title("Median Blur")
plt.axis("off")
plt.show()
```

```python
# =========================================================
# 5. Sharpening Filter - Laplacian Linear Kernel
# =========================================================

kernel2 = np.array([
    [-1, -1, -1],
    [2, -2, 1],
    [2, 1, -1]
])

laplacian_kernel = cv2.filter2D(
    image2,
    -1,
    kernel2
)

plt.imshow(laplacian_kernel)
plt.title("Laplacian Kernel")
plt.axis("off")
plt.show()
```

```python
# =========================================================
# 6. Sharpening Filter - Laplacian Operator
# =========================================================

laplacian = cv2.Laplacian(
    image2,
    cv2.CV_64F
)

plt.imshow(laplacian)
plt.title("Laplacian Operator")
plt.axis("off")
plt.show()
```
